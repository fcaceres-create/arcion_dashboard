Análisis exploratorio inicial del dataset de donantes de sangre en Argentina.
Carga el dataset maestro (sintético hasta tener datos oficiales del Plan Nacional de Sangre)
y revisa distribuciones, correlaciones y patrones temporales.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import cargar_dataset_maestro
from src import config

sns.set_style('whitegrid')
df = cargar_dataset_maestro()
df.head()
```

## Distribución de la tasa por región

```python
df_hist = df[df['Año'] <= config.ANIO_FIN_HISTORICO]
fig, ax = plt.subplots(figsize=(10, 5))
sns.boxplot(data=df_hist, x='Region', y='Tasa_Donacion_x1000', ax=ax)
ax.axhline(config.OMS_OPTIMO_X1000, ls='--', color='orange', label='Meta OMS')
ax.legend()
plt.show()
```

## Evolución temporal de la tasa nacional

```python
tasa_nacional = (
    df_hist.groupby('Año')
    .apply(lambda g: g['Donantes_Anuales'].sum() / g['Población_Total'].sum() * 1000)
    .rename('Tasa_Nacional')
)
tasa_nacional.plot(marker='o', figsize=(10, 4), title='Tasa nacional histórica')
plt.axhline(config.OMS_OPTIMO_X1000, ls='--', color='orange', label='Meta OMS')
plt.legend()
plt.show()
```
