# 02 - Modelado

Pruebas comparativas de modelos: Random Forest vs Gradient Boosting,
validación cruzada y análisis de feature importance.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
from src.data_loader import cargar_dataset_maestro
from src.feature_engineering import construir_features
from src.model import entrenar_y_seleccionar
from src import config

df = cargar_dataset_maestro()
df_hist = df[df['Año'] <= config.ANIO_FIN_HISTORICO]
df_features = construir_features(df_hist)
resultado = entrenar_y_seleccionar(df_features, guardar_metricas=False)

print('Modelo ganador:', resultado.nombre_mejor)
for nombre, m in resultado.metricas.items():
    print(f'  {nombre} | MAE-CV={m.mae_cv:.3f} | R2={m.r2_train:.3f}')
```

## Top 10 features

```python
imp = resultado.metricas[resultado.nombre_mejor].feature_importance
pd.Series(imp).head(10).plot(kind='barh', figsize=(8, 5)).invert_yaxis()
```
