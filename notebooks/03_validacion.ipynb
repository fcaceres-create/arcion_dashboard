# 03 - Validación de proyecciones

Análisis de los outputs del pipeline: chequeo de coherencia,
comparación entre escenarios y verificación de rangos plausibles.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
from src import config

df = pd.read_csv(config.ARCHIVO_PROYECCION_CSV, encoding='utf-8-sig')
print('Filas:', len(df), '| Escenarios:', df['Escenario'].unique())
df.head()
```

## Tasa nacional por escenario

```python
nac = df.groupby(['Año', 'Escenario']).apply(
    lambda g: g['Donantes_Anuales'].sum() / g['Población_Total'].sum() * 1000
).rename('Tasa_Nacional').reset_index()

fig, ax = plt.subplots(figsize=(11, 5))
for esc in nac['Escenario'].unique():
    sub = nac[nac['Escenario'] == esc]
    ax.plot(sub['Año'], sub['Tasa_Nacional'], marker='o', label=esc)
ax.axhline(config.OMS_OPTIMO_X1000, ls='--', color='orange', label='Meta OMS')
ax.set_title('Tasa nacional por escenario')
ax.set_xlabel('Año'); ax.set_ylabel('Donaciones / 1000 hab')
ax.legend()
plt.show()
```
